## Step 3️⃣: Import and Initialize

```python
from three_stream_architecture import ThreeStreamArchitecture
import numpy as np

# Initialize the detector
kg_detector = ThreeStreamArchitecture(
    techniques_path="propaganda_techniques.json",
    use_llm=False  # Set to True if you have OpenAI/Anthropic API key
)

print("✅ Knowledge Graph detector initialized!")
print(f"   Loaded {len(kg_detector.technique_names)} propaganda techniques")
```

## Step 4️⃣: Example - How It Works

Let's see how the Knowledge Graph improves predictions on an example meme.

```python
# Example meme
meme_text = """
These corrupt politicians are lying to you!
Everyone knows the media is controlled.
Don't be a sheep - wake up!
"""

print("Meme text:")
print(meme_text)
```

```python
# SIMULATED neural predictions (replace this with YOUR model)
# This is what your CLIP + RoBERTa model would output

# Create predictions for all techniques
neural_predictions = {tech: 0.05 for tech in kg_detector.technique_names}

# Set some techniques higher (simulating your model's output)
neural_predictions['Loaded Language'] = 0.72
neural_predictions['Appeal to (Strong) Emotions'] = 0.68
neural_predictions['Doubt'] = 0.55
neural_predictions['Bandwagon'] = 0.45
neural_predictions['Name Calling/Labeling'] = 0.38

print("🧠 Neural Model Predictions (Top 5):")
for tech, score in sorted(neural_predictions.items(), key=lambda x: -x[1])[:5]:
    print(f"   {tech}: {score:.3f}")
```

```python
# Apply Knowledge Graph reasoning
result = kg_detector.predict(
    text=meme_text,
    image_features={},  # Empty for now (can add later)
    neural_predictions=neural_predictions,
    return_explanations=True
)

print("\n✨ After Knowledge Graph Reasoning:")
print(f"\nDetected {len(result['detected_techniques'])} techniques:\n")

for tech in result['detected_techniques']:
    neural = result['stream_predictions']['neural'][tech]
    kg = result['stream_predictions']['knowledge_graph'][tech]
    final = result['final_predictions'][tech]
    
    print(f"✓ {tech}")
    print(f"  Neural: {neural:.3f} → KG: {kg:.3f} → Final: {final:.3f}")
    print()
```
